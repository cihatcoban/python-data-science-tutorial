### 1. Merkezi Limit Teoremi (Central Limit Theorem - CLT)
Orijinal veri hangi dağılıma sahip olursa olsun, yeterli büyüklükteki örneklemlerin ortalamaları Normal Dağılıma yaklaşır.


```python
import numpy as np

# Düzgün (Uniform) dağılımdan örneklem çekip ortalamalarını hesaplama
population = np.random.uniform(low=0, high=100, size=100_000)
sample_means = [np.mean(np.random.choice(population, size=50)) for _ in range(1000)]

print(f"Popülasyon Ortalaması   : {np.mean(population):.2f}")
print(f"Örneklem Ortalamaları Ort: {np.mean(sample_means):.2f}")
print(f"Örneklem Standart Hatası: {np.std(sample_means):.2f}")
```

### 2. İki Bağımsız Örneklem $t$-Testi (Student's t-Test)
İki grup arasındaki ortalama farkının şans eseri mi yoksa istatistiksel olarak anlamlı mı olduğunu test eder ($p < 0.05$).


```python
from scipy import stats

group_a = np.random.normal(loc=72, scale=10, size=40)  # Yeni eğitim yöntemi
group_b = np.random.normal(loc=65, scale=12, size=40)  # Geleneksel yöntem

t_stat, p_val = stats.ttest_ind(group_a, group_b)
print(f"t-istatistiği: {t_stat:.4f} | p-değeri: {p_val:.4f}")

if p_val < 0.05:
    print("Sonuç: p < 0.05, Gruplar arasındaki başarı farkı istatistiksel olarak ANLAMLI.")
else:
    print("Sonuç: p >= 0.05, İki grup arasında anlamlı bir fark kanıtlanamadı (H0 reddedilemez).")
```
