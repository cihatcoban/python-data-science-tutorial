### 1. Matplotlib Object-Oriented API Standartları
`plt.plot()` yerine `fig, ax = plt.subplots()` kullanmak kodun genişletilebilirliği ve eksen kontrolü için esastır.


```python
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(42)
x = np.linspace(0, 10, 100)
y = np.sin(x) + np.random.normal(0, 0.2, 100)

fig, ax = plt.subplots(figsize=(8, 4), dpi=100)
ax.plot(x, np.sin(x), label="Gerçek Fonksiyon (sin)", color="#1f77b4", lw=2)
ax.scatter(x, y, label="Gözlemlenen Veri (Gürültülü)", color="#ff7f0e", alpha=0.6, s=20)

ax.set_title("Sinyal ve Gözlem Analizi", fontsize=12, fontweight="bold")
ax.set_xlabel("Zaman / X", fontsize=10)
ax.set_ylabel("Genlik / Y", fontsize=10)
ax.grid(True, linestyle="--", alpha=0.5)
ax.legend(frameon=True)

plt.tight_layout()
plt.show()
```

### 2. Korelasyon Analizi ve Isı Haritası (Heatmap)
Değişkenler arasındaki doğrusal ilişki katsayılarını görselleştirme.


```python
import pandas as pd

housing_df = pd.read_csv("../../data/raw/sample_housing.csv")
numeric_df = housing_df.select_dtypes(include=[np.number])
corr = numeric_df.corr()

print("Korelasyon Matrisi (Fiyat İlişkisi):")
print(corr["price"].sort_values(ascending=False).round(3))
```
