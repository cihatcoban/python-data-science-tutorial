### 1. Orijinal Scraping Kodlarının Modernizasyonu: Etik ve Güvenli HTML Ayrıştırma
Web scraping yaparken siteleri DoS etmemek için `headers` eklemek ve istek aralarına bekleme süreleri koymak esastır.


```python
from bs4 import BeautifulSoup

# Güvenli HTML parsing örneği
sample_html = '''
<html>
  <body>
    <div class="product-card" data-id="101">
      <h2 class="title">Veri Bilimi Kursu</h2>
      <span class="price">499.99</span>
      <span class="rating">4.8</span>
    </div>
    <div class="product-card" data-id="102">
      <h2 class="title">Makine Öğrenmesi Kitabı</h2>
      <span class="price">189.50</span>
      <span class="rating">4.6</span>
    </div>
  </body>
</html>
'''

soup = BeautifulSoup(sample_html, "html.parser")
items = []
for card in soup.find_all("div", class_="product-card"):
    title = card.find("h2", class_="title").text.strip()
    price = float(card.find("span", class_="price").text)
    rating = float(card.find("span", class_="rating").text)
    items.append({"title": title, "price": price, "rating": rating})

import pandas as pd
df_scraped = pd.DataFrame(items)
print("Ayrıştırılan Ürün Tablosu:")
print(df_scraped)
```
