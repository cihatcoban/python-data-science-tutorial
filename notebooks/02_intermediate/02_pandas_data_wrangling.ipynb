### 1. DataFrame Oluşturma ve İndeksleme Mantığı (.loc vs .iloc)
- `.loc`: Etiket (label) tabanlı seçim yapar.
- `.iloc`: Tamsayı pozisyon (integer index) tabanlı seçim yapar.


```python
import pandas as pd
import numpy as np

df = pd.read_csv("../../data/raw/sample_students.csv")
print("İlk 5 Satır:")
print(df.head())

# .loc ile filtreleme: Sadece Bilgisayar Mühendisliği öğrencileri ve final notları
cs_students = df.loc[df["department"] == "Computer Science", ["name", "final_exam"]]
print("\nBilgisayar Mühendisliği Final Notları:")
print(cs_students)
```

### 2. GroupBy ve Çoklu Agregasyon (Multi-Aggregation)
Departman bazında ortalama sınav notları ve katılım oranlarını inceleme.


```python
dept_summary = df.groupby("department").agg(
    ogrenci_sayisi=("student_id", "count"),
    vize1_ortalama=("midterm1", "mean"),
    final_ortalama=("final_exam", "mean"),
    katilim_ort=("attendance_rate", "mean")
).round(2)

print("Departman Karnesi:")
print(dept_summary)
```
