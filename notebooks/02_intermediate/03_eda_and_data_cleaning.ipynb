### 1. Eksik Değer (Missing Value) Stratejileri
Gerçek hayat verilerinde eksik değerleri körlemesine silmek veri kaybına ve yanlılığa (bias) yol açar.


```python
import pandas as pd
import numpy as np

# Simüle edilmiş kirli veri kümesi
dirty_data = pd.DataFrame({
    "age": [25, 30, np.nan, 45, 120, 22],  # 120 aykırı değer, NaN eksik
    "income": [45000, 54000, 60000, np.nan, 52000, 48000],
    "city": ["Istanbul", "Ankara", np.nan, "Izmir", "Istanbul", "Bursa"]
})

print("Eksik Değer Sayıları:")
print(dirty_data.isnull().sum())

# Sayısal sütunları medyan ile, kategorik sütunları mod ile doldurma
dirty_data["age"] = dirty_data["age"].fillna(dirty_data["age"].median())
dirty_data["income"] = dirty_data["income"].fillna(dirty_data["income"].median())
dirty_data["city"] = dirty_data["city"].fillna(dirty_data["city"].mode()[0])
print("\nDoldurulmuş Veri:")
print(dirty_data)
```

### 2. Aykırı Değer (Outlier) Tespiti: Tukey's IQR Yöntemi
$IQR = Q3 - Q1$, Sınırlar: $[Q1 - 1.5 \times IQR, Q3 + 1.5 \times IQR]$.


```python
q1 = dirty_data["age"].quantile(0.25)
q3 = dirty_data["age"].quantile(0.75)
iqr = q3 - q1
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

outliers = dirty_data[(dirty_data["age"] < lower_bound) | (dirty_data["age"] > upper_bound)]
print(f"IQR: {iqr:.2f} | Alt Sınır: {lower_bound:.2f} | Üst Sınır: {upper_bound:.2f}")
print("Aykırı Kayıtlar:")
print(outliers)
```
