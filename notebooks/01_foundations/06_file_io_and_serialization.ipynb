### 1. Orijinal Kurs Projesinin Modernizasyonu: Öğrenci Not Analiz Sistemi
Eski kurstaki ham metin tabanlı not hesaplama mantığını, CSV desteği ve tip kontrollü modern bir pipeline haline getiriyoruz.


```python
import csv
from pathlib import Path

def calculate_letter_grade(vize1: float, vize2: float, final: float) -> tuple[float, str]:
    total = (vize1 * 0.3) + (vize2 * 0.3) + (final * 0.4)
    if total >= 90:
        return total, "AA"
    elif total >= 80:
        return total, "BA"
    elif total >= 70:
        return total, "BB"
    elif total >= 60:
        return total, "CB"
    elif total >= 50:
        return total, "CC"
    else:
        return total, "FF"

# data/raw/sample_students.csv dosyasını okuyup notları hesaplayalım
input_file = Path("../../data/raw/sample_students.csv")
if not input_file.exists():
    input_file = Path("data/raw/sample_students.csv")

print(f"Reading from: {input_file}")
with open(input_file, mode="r", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    print(f"{'Öğrenci Adı':<18} | {'Ortalama':<8} | {'Harf Notu':<8} | {'Durum'}")
    print("-" * 50)
    for row in list(reader)[:6]:
        avg, letter = calculate_letter_grade(
            float(row["midterm1"]), float(row["midterm2"]), float(row["final_exam"])
        )
        status = "GEÇTİ" if letter != "FF" else "KALDI"
        print(f"{row['name']:<18} | {avg:<8.2f} | {letter:<8} | {status}")
```
