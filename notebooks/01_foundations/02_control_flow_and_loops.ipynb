### 1. Karar Blokları ve Eşik Değeri Sınıflandırması
Veri biliminde model tahminlerini olasılıktan sınıfa dönüştürürken veya veri kalite kontrollerinde koşullu ifadeler temeldir.


```python
def classify_risk(probability: float) -> str:
    if not (0.0 <= probability <= 1.0):
        raise ValueError(f"Olasılık [0, 1] aralığında olmalıdır, girilen: {probability}")
    
    if probability >= 0.8:
        return "YÜKSEK_RİSK"
    elif probability >= 0.4:
        return "ORTA_RİSK"
    else:
        return "DÜŞÜK_RİSK"

sample_probs = [0.15, 0.45, 0.88, 0.72]
for prob in sample_probs:
    print(f"Prob: {prob:.2f} -> Sınıf: {classify_risk(prob)}")
```

### 2. İki Listeyi Eşzamanlı İşleme: `zip` ve İndeksleme: `enumerate`
`range(len(...))` yazmak yerine `zip` ve `enumerate` kullanmak Pythonic standarttır.


```python
true_labels = [1, 0, 1, 1, 0, 1, 0, 0]
pred_labels = [1, 0, 1, 0, 0, 1, 1, 0]

# Doğru tahminlerin tespiti
matches = 0
for idx, (y_true, y_pred) in enumerate(zip(true_labels, pred_labels)):
    is_correct = (y_true == y_pred)
    if is_correct:
        matches += 1
    print(f"Örnek {idx+1}: Gerçek={y_true}, Tahmin={y_pred} | {'DOĞRU' if is_correct else 'HATALI'}")

accuracy = matches / len(true_labels)
print(f"\nDoğruluk (Accuracy): {accuracy:.2%}")
```
