### 1. Klasik OOP: Model ve Metrik Takip Sınıfı
OOP, karmaşık makine öğrenmesi iş akışlarını kapsüllemek (encapsulate) için endüstri standardıdır.


```python
class MetricTracker:
    def __init__(self, experiment_name: str):
        self.experiment_name = experiment_name
        self.history: dict[str, list[float]] = {}

    def log(self, metric_name: str, value: float) -> None:
        if metric_name not in self.history:
            self.history[metric_name] = []
        self.history[metric_name].append(value)

    def summary(self) -> dict[str, float]:
        return {k: sum(v) / len(v) for k, v in self.history.items() if v}

    def __repr__(self) -> str:
        return f"<MetricTracker: {self.experiment_name} | Tracked: {list(self.history.keys())}>"

tracker = MetricTracker("LogisticRegression_V1")
tracker.log("loss", 0.65)
tracker.log("loss", 0.42)
tracker.log("loss", 0.31)
tracker.log("accuracy", 0.88)
print(tracker)
print("Ortalamalar:", tracker.summary())
```

### 2. Modern Python: `@dataclass` ile Temiz Veri Mimarisi
Boilerplate `__init__` kodlarını ortadan kaldıran ve `dataclasses` modülü ile oluşturulan şemalar.


```python
from dataclasses import dataclass, field
from datetime import datetime

@dataclass
class DatasetSchema:
    name: str
    num_rows: int
    num_features: int
    has_target: bool = True
    created_at: str = field(default_factory=lambda: datetime.now().strftime("%Y-%m-%d"))

    @property
    def total_cells(self) -> int:
        return self.num_rows * self.num_features

schema = DatasetSchema(name="credit_churn_v1", num_rows=10_000, num_features=25)
print(schema)
print(f"Toplam Hücre Sayısı: {schema.total_cells:,}")
```
