### Egzersiz 1: Fibonacci Dizisi ve Jeneratör Implementasyonu
Verilen $N$ adımına kadar Fibonacci serisini bellek tüketmeden üreten fonksiyon.


```python
def fibonacci_generator(limit: int):
    a, b = 0, 1
    count = 0
    while count < limit:
        yield a
        a, b = b, a + b
        count += 1

print("İlk 10 Fibonacci sayısı:", list(fibonacci_generator(10)))
```

### Egzersiz 2: Asal Sayı Doğrulama ve Eratosthenes Eleği


```python
def is_prime(n: int) -> bool:
    if n <= 1:
        return False
    if n <= 3:
        return True
    if n % 2 == 0 or n % 3 == 0:
        return False
    i = 5
    while i * i <= n:
        if n % i == 0 or n % (i + 2) == 0:
            return False
        i += 6
    return True

test_nums = [2, 3, 17, 24, 97, 100]
print({x: is_prime(x) for x in test_nums})
```

### Egzersiz 3: Nesne Tabanlı ATM Hesap Yöneticisi


```python
class BankAccount:
    def __init__(self, owner: str, initial_balance: float = 0.0):
        self.owner = owner
        self._balance = initial_balance

    def deposit(self, amount: float) -> float:
        if amount <= 0:
            raise ValueError("Yatırılacak tutar pozitif olmalıdır.")
        self._balance += amount
        return self._balance

    def withdraw(self, amount: float) -> float:
        if amount > self._balance:
            raise ValueError("Yetersiz bakiye.")
        self._balance -= amount
        return self._balance

    @property
    def balance(self) -> float:
        return self._balance

acc = BankAccount("Cihat Çoban", 5000)
acc.deposit(1500)
acc.withdraw(2000)
print(f"Hesap Sahibi: {acc.owner} | Güncel Bakiye: ₺{acc.balance:,.2f}")
```
