### 1. Tip İpuçları (Type Hinting) ve Esnek Parametreler
Keystone standardı gereğince fonksiyonların girdi ve çıktı tipleri açıkça belirtilir.


```python
from typing import List, Dict, Any, Callable
import time

def min_max_scaler(values: List[float]) -> List[float]:
    """Verilen sayı listesini [0, 1] aralığına normalize eder."""
    if not values:
        return []
    min_v, max_v = min(values), max(values)
    if min_v == max_v:
        return [0.0] * len(values)
    return [(x - min_v) / (max_v - min_v) for x in values]

raw_data = [10.0, 20.0, 50.0, 100.0]
print("Ölçeklenmiş Veri:", min_max_scaler(raw_data))
```

### 2. Fonksiyon Çalışma Süresi Ölçümü İçin Dekoratör (Decorator) Tasarımı
Büyük veri işleme adımlarında performans profili çıkarmak için dekoratörler vazgeçilmezdir.


```python
def time_it(func: Callable) -> Callable:
    """Fonksiyonun çalışma süresini milisaniye cinsinden ölçen dekoratör."""
    def wrapper(*args, **kwargs):
        start = time.perf_counter()
        result = func(*args, **kwargs)
        duration_ms = (time.perf_counter() - start) * 1000
        print(f"[PERF] '{func.__name__}' tamamlandı: {duration_ms:.3f} ms")
        return result
    return wrapper

@time_it
def compute_heavy_simulation(n: int) -> float:
    return sum(x ** 0.5 for x in range(n))

val = compute_heavy_simulation(100_000)
```

### 3. Jeneratörler (Generators) ile Hafıza Dostu Veri Akışı
Tüm veri kümesini RAM'e yüklemek yerine `yield` kullanarak satır satır akış sağlanır.


```python
def stream_batch_data(total_samples: int, batch_size: int):
    """Bellek taşmasını önleyen batch veri üreticisi."""
    current = 0
    while current < total_samples:
        batch = list(range(current, min(current + batch_size, total_samples)))
        yield batch
        current += batch_size

for b_idx, batch in enumerate(stream_batch_data(total_samples=25, batch_size=10)):
    print(f"Batch #{b_idx+1}: {batch}")
```
