### 1. Savunmacı Programlama ve Hata Yakalama
Veri hatları beklenmeyen eksik değerler, bozuk formatlar veya sıfıra bölme hataları içerebilir. Hataları sessizce yutmak yerine kontrollü şekilde yakalamalıyız.


```python
import logging

logging.basicConfig(level=logging.INFO, format="%(levelname)s: %(message)s")

class DataValidationError(Exception):
    """Veri şeması veya içeriği geçersiz olduğunda fırlatılır."""
    pass

def validate_dataframe_columns(actual_cols: list[str], expected_cols: list[str]) -> bool:
    missing = set(expected_cols) - set(actual_cols)
    if missing:
        raise DataValidationError(f"Zorunlu sütunlar eksik: {missing}")
    return True

try:
    cols = ["id", "age", "income"]
    required = ["id", "age", "income", "target"]
    validate_dataframe_columns(cols, required)
except DataValidationError as e:
    logging.error(f"Validasyon başarısız oldu! Detay: {e}")
else:
    logging.info("Veri şeması başarıyla doğrulandı.")
finally:
    logging.info("Doğrulama adımı tamamlandı.")
```
