### 2. Koleksiyonlar: List, Tuple, Dict ve Set Karşılaştırması
Veri biliminde hangi veri yapısının seçileceği doğrudan performans ($O(1)$ vs $O(n)$) etkiler.
- **List**: Değiştirilebilir (mutable), sıralı.
- **Tuple**: Değiştirilemez (immutable), hafif ve hashable.
- **Dict**: Anahtar-değer haritası, $O(1)$ ortalama erişim süresi.
- **Set**: Benzersiz elemanlar, küme operasyonları (kesişim, birleşim).


```python
# Veri Bilimi Örneği: Feature Adları ve Konfigürasyon
features_list = ["age", "income", "credit_score", "age"]
features_set = set(features_list)  # Benzersiz özellikler
print(f"Orijinal liste uzunluğu: {len(features_list)}, Tekil özellik sayısı: {len(features_set)}")

# Tuple: Sabit hiperparametreler
image_dimensions = (224, 224, 3)  # Height, Width, Channels

# Dict: Model metadata
metadata = {
    "model_id": "RF_001",
    "n_estimators": 100,
    "max_depth": 10,
    "metrics": {"f1": 0.89, "accuracy": 0.92}
}
print(f"Model F1-Skoru: {metadata['metrics']['f1']}")
```

### 3. List Comprehensions ve Generator Expressions
List comprehension, verileri filtrelemek ve dönüştürmek için döngülere kıyasla hem daha okunaklı hem de CPython seviyesinde daha hızlıdır.


```python
raw_measurements = [12.4, -999.0, 15.6, 18.2, -999.0, 22.1]

# Hatalı değerleri (-999.0) filtreleyip log dönüşümü simülasyonu
clean_measurements = [x for x in raw_measurements if x > 0]
print("Temiz Ölçümler:", clean_measurements)

# Dict Comprehension: Özellik indeks haritası
feature_names = ["age", "balance", "tenure", "num_products"]
feature_to_idx = {name: idx for idx, name in enumerate(feature_names)}
print("Özellik İndeks Haritası:", feature_to_idx)
```
