### 1. Sayısal Özellik Ölçekleme: Standart vs Robust Ölçekleyici
Uç değerlerin (outliers) bulunduğu veri setlerinde `StandardScaler` ortalama ve standart sapmayı bozacağı için `RobustScaler` (medyan ve IQR tabanlı) tercih edilmelidir.


```python
import numpy as np
import pandas as pd

# Sayısal veri simülasyonu (outlier içeren)
vals = np.array([10.0, 12.0, 11.5, 13.0, 9.8, 100.0])  # 100 aşırı aykırı değer

# Z-Score Standardizasyonu
z_scaled = (vals - np.mean(vals)) / np.std(vals)

# Robust Ölçekleme (Medyan ve IQR)
median = np.median(vals)
q75, q25 = np.percentile(vals, [75, 25])
iqr = q75 - q25
robust_scaled = (vals - median) / iqr

df_scale = pd.DataFrame({"Original": vals, "Standard_Z": z_scaled.round(2), "Robust_IQR": robust_scaled.round(2)})
print(df_scale)
```

### 2. Kategorik Değişken Kodlama (Encoding)
- **One-Hot Encoding**: Sıralama ilişkisi olmayan nominal kategoriler için.
- **Ordinal Encoding**: 'Düşük', 'Orta', 'Yüksek' gibi doğal sıralaması olan ordinal veriler için.


```python
df_cat = pd.DataFrame({
    "department": ["Computer Science", "Data Engineering", "Statistics", "Computer Science"],
    "seniority": ["Junior", "Senior", "Mid", "Junior"]
})

# One-Hot Encoding
one_hot = pd.get_dummies(df_cat["department"], prefix="dept", dtype=int)

# Ordinal Encoding Sözlüğü
seniority_map = {"Junior": 1, "Mid": 2, "Senior": 3}
df_cat["seniority_encoded"] = df_cat["seniority"].map(seniority_map)

encoded_df = pd.concat([df_cat[["seniority_encoded"]], one_hot], axis=1)
print("Kodlanmış Özellik Tablosu:")
print(encoded_df)
```
