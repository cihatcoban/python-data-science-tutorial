### 1. K-Means Kümeleme ve Müşteri/Öğrenci Segmentasyonu
Etiketsiz verilerde doğal grupları ve paternleri keşfetmek için mesafe tabanlı kümeleme.


```python
import numpy as np
import pandas as pd

np.random.seed(42)
# 3 Farklı Öğrenci Segmenti Simülasyonu (Vize vs Final Notları)
cluster_1 = np.random.normal(loc=[85, 90], scale=5, size=(30, 2))  # Başarılı grup
cluster_2 = np.random.normal(loc=[55, 60], scale=6, size=(30, 2))  # Orta grup
cluster_3 = np.random.normal(loc=[35, 40], scale=5, size=(30, 2))  # Destek gereken grup

X_synth = np.vstack([cluster_1, cluster_2, cluster_3])
print(f"Sentetik Kümeleme Veri Boyutu: {X_synth.shape}")
```

### 2. Boyut İndirgeme (PCA) ve Varyans Korunumu


```python
# Özdeğer / Özvektör ile 2D -> 1D İndirgeme
X_centered = X_synth - np.mean(X_synth, axis=0)
cov_mat = np.cov(X_centered, rowvar=False)
eigenvalues, eigenvectors = np.linalg.eigh(cov_mat)

# En büyük özdeğere sahip vektör
idx_max = np.argmax(eigenvalues)
pc1 = eigenvectors[:, idx_max]
explained_var_ratio = eigenvalues[idx_max] / np.sum(eigenvalues)

print(f"1. Asıl Bileşenin Koruduğu Varyans Oranı: %{explained_var_ratio * 100:.2f}")
```
