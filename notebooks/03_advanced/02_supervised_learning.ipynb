### 1. Konut Fiyat Tahmini: Regresyon Modellemesi
`data/raw/sample_housing.csv` verisi üzerinden bağımsız değişkenlerle hedef değişken (fiyat) arasındaki ilişkiyi kurma.


```python
import pandas as pd
import numpy as np

df = pd.read_csv("../../data/raw/sample_housing.csv")
features = ["area_sqm", "rooms", "building_age", "distance_to_center_km", "has_parking"]
target = "price"

X = df[features]
y = df[target]

print(f"X Boyutu: {X.shape}, y Boyutu: {y.shape}")

# OLS Çoklu Lineer Regresyon Çözümü (Normal Denklem: theta = (X^T X)^-1 X^T y)
X_b = np.c_[np.ones((len(X), 1)), X.values]  # Bias terimi ekleme
theta_best = np.linalg.pinv(X_b.T.dot(X_b)).dot(X_b.T).dot(y.values)

print("Hesaplanan Model Ağırlıkları (Weights):")
print(f"Bias (Intercept) : ₺{theta_best[0]:,.2f}")
for feat, coef in zip(features, theta_best[1:]):
    print(f"{feat:<22}: {coef:,.2f}")
```

### 2. Model Tahminleri ve Basit Performans Değerlendirmesi


```python
y_pred = X_b.dot(theta_best)
mae = np.mean(np.abs(y.values - y_pred))
rmse = np.sqrt(np.mean((y.values - y_pred)**2))
r2 = 1 - (np.sum((y.values - y_pred)**2) / np.sum((y.values - np.mean(y.values))**2))

print(f"MAE  (Ortalama Mutlak Hata): ₺{mae:,.2f}")
print(f"RMSE (Kök Ortalama Kare Hata): ₺{rmse:,.2f}")
print(f"R^2  (Belirlilik Katsayısı)   : {r2:.4f}")
```
